```python
#Step 4 Comprehensive EDA Analysis (Bivariate Analysis)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('/content/student_habits_performance.xlsx')
# Drop rows with missing parental education level for plotting
df_cleaned = df.dropna(subset=['parental_education_level']).copy()

# Group by parental education level and count the occurrences
education_counts = df_cleaned['parental_education_level'].value_counts().sort_index()

x = education_counts.index
y = education_counts.values

plt.bar(x,y)
plt.xlabel('Parental Education Level')
plt.ylabel('Number of Students')
plt.title('Number of Students by Parental Education Level')
plt.show()
```

```python
#Extra Analysis 2
data = pd.read_excel('/content/student_habits_performance.xlsx')

# Filter data for students with exam scores over 70
high_score_students = data[data['exam_score'] > 70].copy()

# Group by diet quality and count students with high scores
high_score_counts = high_score_students['diet_quality'].value_counts().sort_index()

# Create the bar plot
ax = sns.barplot(x=high_score_counts.index, y=high_score_counts.values, ci=None) # ci=None to avoid confidence intervals

# Add callouts to the bars
for i, patch in enumerate(ax.patches):
    ax.annotate(f'{high_score_counts.values[i]:.0f}', (patch.get_x() + patch.get_width() / 2., patch.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points')


plt.xlabel('Diet Quality')
plt.ylabel('Number of Students with Exam Score > 70') # Set the y-axis label
plt.title('Number of Students with Exam Score > 70 by Diet Quality') # Add a title

plt.show()
```
